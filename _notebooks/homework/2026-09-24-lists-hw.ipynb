## Popcorn Hack 1 — Predict, run, fix the crash

The pseudocode I was converting:

    foods <- ["pizza", "sushi", "tacos", "ramen"]

    DISPLAY(foods[1])
    DISPLAY(foods[LENGTH(foods)])
    DISPLAY(LENGTH(foods))
    DISPLAY(foods[5])

**The off-by-one the whole hack is about.** AP CSP pseudocode starts counting at 1, Python starts at 0. So the same four elements have valid indexes **1 through 4** in pseudocode and **0 through 3** in Python. Every line has to shift down by one on the way across.

My predictions before running, in order: `pizza`, `ramen`, `4`, and a crash.

**Part 2 — the error.** The fourth line was the crash. `foods[5]` does not exist in either language: pseudocode tops out at `foods[4]` and Python tops out at `foods[3]`. Python raises `IndexError: list index out of range`. My fix was `foods[3]`, which is the Python translation of the pseudocode's `foods[4]` — the last element, found with `len(foods) - 1` rather than typed in as a number.


```python
# CODE_RUNNER: Lists Popcorn 1 - predict, run, fix the crash

foods = ["pizza", "sushi", "tacos", "ramen"]

# DISPLAY(foods[1])  ->  pseudocode index 1 is Python index 0
print(foods[0])                    # prediction: pizza

# DISPLAY(foods[LENGTH(foods)])  ->  last element, so len() - 1 in Python
print(foods[len(foods) - 1])       # prediction: ramen

# DISPLAY(LENGTH(foods))
print(len(foods))                  # prediction: 4

# DISPLAY(foods[5])  ->  crashed: valid Python indexes are 0 through 3
# print(foods[5])                  # IndexError: list index out of range
print(foods[3])                    # fixed: last valid index
```

## Popcorn Hack 2 — Sum and average with an accumulator

Six numbers called out in class. The accumulator pattern is three steps and the order is the whole point:

1. `total = 0` **before** the loop. Put it inside and it resets on every pass, so you end up with just the last number.
2. `total = total + number` **inside** the loop.
3. Use `total` **after** the loop finishes.

The second trap is the average. `total / len(numbers)` raises `ZeroDivisionError` the moment the list is empty, so I guarded it with `if len(numbers) == 0` instead of assuming data exists. Expected: total 42, average 7.0.


```python
# CODE_RUNNER: Lists Popcorn 2 - accumulator sum and average

numbers = [7, 6, 8, 5, 7, 9]

# 1. initialize BEFORE the loop
total = 0

# 2. add each element INSIDE the loop
for number in numbers:
    total = total + number

# 3. use it AFTER the loop
print("Total:", total)

# guard against the empty list before dividing
if len(numbers) == 0:
    print("No data yet")
else:
    average = total / len(numbers)
    print("Average:", round(average, 2))
```

## Popcorn Hack 3 — Count the class survey, then filter it

Two patterns on one loop each.

**Frequency count.** One dictionary handles any number of categories, including ones you have never seen before. For each item: if the key is already there, add 1; otherwise start it at 1. The alternative is a separate counter variable per season, which stops working the second somebody answers with a season you did not plan for.

**Filter.** Build a *new* list and leave the original alone. Removing items from a list while looping over it makes the loop skip elements, because the positions shift underneath it.

Expected: winter 1, summer 3, fall 2, spring 1 — and `['summer', 'fall']` once I keep only the seasons with two or more votes.


```python
# CODE_RUNNER: Lists Popcorn 3 - count then filter

seasons = ["winter", "summer", "fall", "summer", "spring", "summer", "fall"]   # class responses

frequency = {}
for season in seasons:
    if season in frequency:
        frequency[season] = frequency[season] + 1
    else:
        frequency[season] = 1

print(frequency)

# 1. Build a list of seasons with 2 or more votes
popular = []
for season in frequency:
    if frequency[season] >= 2:
        popular.append(season)

# 2. Print that list
print(popular)

# the original list is untouched - the filter built a new one
print(seasons)
```

## Homework Hack — all three lessons on one dataset

Snack votes from the class party, and what each snack cost.

- **Lesson 1, indexing.** `len(votes)` for the count, and `votes[len(votes) - 1]` for the last vote. Not a hardcoded `11` — that breaks the moment one more person votes.
- **Lesson 2, accumulator.** `total = 0` before the loop, `total = total + price` inside it, no `sum()`.
- **Lesson 3, count and filter.** One dictionary for the counts, then a second loop that appends the snacks clearing the threshold into a new list.

Expected output shape:

    12
    cookies
    Total: 31
    Average: 2
    {'chips': 5, 'cookies': 4, 'fruit': 2, 'pretzels': 1}
    ['chips', 'cookies']

**Why the average prints 2 and not 2.58.** 31 divided by 12 is 2.583..., and `int()` truncates toward zero rather than rounding — it drops the decimal instead of rounding it up. That is what the expected output asks for, but it does understate the real average by more than half a cent per snack, which is the kind of quiet rounding choice worth noticing before it ends up in something that matters.


```python
# CODE_RUNNER: Lists Homework - indexing, accumulator, count and filter

votes = ["chips", "cookies", "chips", "fruit", "cookies", "chips",
         "pretzels", "cookies", "chips", "fruit", "chips", "cookies"]

prices = [2, 3, 2, 4, 3, 2, 1, 3, 2, 4, 2, 3]

# LESSON 1 - indexing
# 1. how many votes were cast
print(len(votes))

# 2. the last vote, found by index rather than typed in
print(votes[len(votes) - 1])

# LESSON 2 - sum and average
# 3. total the prices with a loop, no sum()
total = 0
for price in prices:
    total = total + price

# 4. print the total and the average as integers
average = total / len(prices)
print("Total:", int(total))
print("Average:", int(average))

# LESSON 3 - count and filter
# 5. frequency dictionary of the votes
frequency = {}
for vote in votes:
    if vote in frequency:
        frequency[vote] = frequency[vote] + 1
    else:
        frequency[vote] = 1

print(frequency)

# 6. snacks with 3 or more votes
popular = []
for snack in frequency:
    if frequency[snack] >= 3:
        popular.append(snack)

print(popular)
```

## What I took away

The three patterns are not interchangeable, and picking the wrong one is most of the difficulty:

| I need | Pattern | Tell |
|---|---|---|
| every value, in order, once | accumulator loop | one running variable initialized before the loop |
| how often something repeats | frequency dictionary | keys you did not know in advance |
| the subset matching a rule | filter into a new list | the original list survives |

The two bugs I would actually hit again are the off-by-one when translating pseudocode to Python, and dividing by `len()` on a list that turned out to be empty. Both are silent until they aren't — the first gives you the wrong element, the second crashes.
