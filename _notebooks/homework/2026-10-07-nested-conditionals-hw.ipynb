## 3.07 Nested Conditionals: Student Life

Homework for the [3.07 Nested Conditionals lesson](https://pages.opencodingsociety.com/python/nested-conditionals/student-life). A **nested conditional** is an `if` inside another `if`, `elif`, or `else` block. The outer decision picks a path first, and the inner decision only gets checked if Python actually walks into the block that holds it. Indentation is what decides which `if` an inner branch belongs to.

Everything here follows one theme from the lesson: a high school student's accounts, homework, meals, and sports.

## Popcorn Hack 1: After-School Homework Plan (Beginner)

**1. Prediction.** With `is_weekday = True` and `homework_finished = False`, it prints **`Finish homework`**.

**2. Path followed.** The **outer `if is_weekday:`** branch is True, so Python goes into that block. Inside it, the **inner `if homework_finished:`** is False, so it takes the **inner `else`** and prints `Finish homework`. The outer `else` (`Weekend plan`) is skipped.


```python
# CODE_RUNNER: Popcorn Hack 1 - Test the after-school homework plan

is_weekday = True
homework_finished = False

if is_weekday:
    if homework_finished:
        print("Free time")
    else:
        print("Finish homework")
else:
    print("Weekend plan")
```

**Actual output:** `Finish homework`, which matches the prediction.

**3. Reaching the other two outcomes.** The cell below runs the same nested conditional for every combination of the two values.


```python
# CODE_RUNNER: Popcorn Hack 1 - Every path of the homework plan

for is_weekday in [True, False]:
    for homework_finished in [True, False]:
        print("is_weekday =", is_weekday, "| homework_finished =", homework_finished, "->", end=" ")
        if is_weekday:
            if homework_finished:
                print("Free time")
            else:
                print("Finish homework")
        else:
            print("Weekend plan")
```

**Actual output:**

| is_weekday | homework_finished | Prints |
|---|---|---|
| True | True | Free time |
| True | False | Finish homework |
| False | True | Weekend plan |
| False | False | Weekend plan |

- **Free time:** `is_weekday = True`, `homework_finished = True`.
- **Weekend plan:** `is_weekday = False`. The value of `homework_finished` does not matter here, because the inner check lives inside the weekday block and is never reached on a weekend.

Two variables give 4 combinations but only **3 outcomes**: both weekend rows land on the same message.


## Popcorn Hack 2: Repair the Student Account Check (Intermediate)

The original code from the lesson:

    logged_in = True
    has_permission = False

    if logged_in:
        print("Account found")
    if has_permission:
        print("Page opened")
    else:
        print("Permission needed")

**1. Why it is not nested.** `if has_permission:` starts at the same indentation as `if logged_in:`, so it is a **separate** `if` statement, not part of the logged-in block. Python checks permission every time, whether the student logged in or not. With `logged_in = False` and `has_permission = True`, the original prints `Page opened`, so a page opens for someone who never logged in. The cell below shows that.


```python
# CODE_RUNNER: Popcorn Hack 2 - The original bug with a logged-out student

logged_in = False
has_permission = True

if logged_in:
    print("Account found")
if has_permission:
    print("Page opened")
else:
    print("Permission needed")
```

**Actual output:** `Page opened`. The student is logged out but still gets in.

**2 and 3. The fix.** The permission check moves one level in, inside the logged-in branch, and an outer `else` prints `Log in first`.


```python
# CODE_RUNNER: Popcorn Hack 2 - Repair the student account check

logged_in = True
has_permission = False

if logged_in:
    print("Account found")
    if has_permission:
        print("Page opened")
    else:
        print("Permission needed")
else:
    print("Log in first")
```

**Actual output:** `Account found`, then `Permission needed`.

**4. Testing all three messages.**


```python
# CODE_RUNNER: Popcorn Hack 2 - Every path of the account check

for logged_in in [True, False]:
    for has_permission in [True, False]:
        print("logged_in =", logged_in, "| has_permission =", has_permission)
        if logged_in:
            print("  Account found")
            if has_permission:
                print("  Page opened")
            else:
                print("  Permission needed")
        else:
            print("  Log in first")
```

**Actual output:**

| logged_in | has_permission | Prints |
|---|---|---|
| True | True | Account found, Page opened |
| True | False | Account found, Permission needed |
| False | True | Log in first |
| False | False | Log in first |

All three messages show up (`Page opened`, `Permission needed`, `Log in first`). The bug from step 1 is gone: logged out with `has_permission = True` now prints `Log in first`, because the permission check can only run inside the logged-in block.


## Popcorn Hack 3: Lunch and Sports Practice (Advanced)

Both TODOs get the same inner question, `practice_after_school`, but each one sits under a different lunch answer, so each inner branch gives its own recommendation. That makes 2 outer paths x 2 inner paths = **4 recommendations**.

**Prediction** for the starting values (`lunch_packed = True`, `practice_after_school = True`): `Eat your packed lunch and bring an extra snack for practice`.


```python
# CODE_RUNNER: Popcorn Hack 3 - Test the lunch and sports-practice plan

lunch_packed = True
practice_after_school = True

if lunch_packed:
    if practice_after_school:
        print("Eat your packed lunch and bring an extra snack for practice")
    else:
        print("Eat your packed lunch - no extra snack needed")
else:
    if practice_after_school:
        print("Buy lunch at school and buy an extra snack for practice")
    else:
        print("Buy lunch at school - no extra snack needed")
```

**Actual output:** `Eat your packed lunch and bring an extra snack for practice`, which matches the prediction.

**Testing all four paths:**


```python
# CODE_RUNNER: Popcorn Hack 3 - Every path of the lunch plan

for lunch_packed in [True, False]:
    for practice_after_school in [True, False]:
        print("lunch_packed =", lunch_packed, "| practice =", practice_after_school, "->", end=" ")
        if lunch_packed:
            if practice_after_school:
                print("Eat your packed lunch and bring an extra snack for practice")
            else:
                print("Eat your packed lunch - no extra snack needed")
        else:
            if practice_after_school:
                print("Buy lunch at school and buy an extra snack for practice")
            else:
                print("Buy lunch at school - no extra snack needed")
```

**Actual output:**

| lunch_packed | practice_after_school | Prints |
|---|---|---|
| True | True | Eat your packed lunch and bring an extra snack for practice |
| True | False | Eat your packed lunch - no extra snack needed |
| False | True | Buy lunch at school and buy an extra snack for practice |
| False | False | Buy lunch at school - no extra snack needed |

Four different recommendations, one per combination.

## Homework Hack: High School Daily Routine Dashboard

**Plan.** The lesson's starter data has four values: login status, homework status, a meal decision (`lunch_packed`), and sports practice. I kept the starter data as given and built the decisions around it.

- **Outer decision:** `logged_in`. If the student is not logged in, nothing else is checked and the dashboard prints `Log in first to see today's plan`.
- **Nested decision 1 (homework):** inside the logged-in block, check `homework_finished`, then inside each answer check `practice_today`. Practice comes before dinner, so it changes when homework happens.
- **Nested decision 2 (lunch):** still inside the logged-in block, check `lunch_packed`, then inside each answer check `practice_today` to decide on an extra snack.

That gives **9 recommendations**: 4 homework messages, 4 lunch messages, and 1 logged-out message. A logged-in student always gets exactly one homework message and one lunch message.

**Prediction** for the starter data (logged in, homework not finished, lunch packed, practice today): `Welcome back - here is today's plan`, then `Go to practice, then finish homework before dinner`, then `Bring your packed lunch and an extra snack for practice`.


```python
# CODE_RUNNER: Homework Hack - Build a high school daily routine dashboard

student = {
    "logged_in": True,
    "homework_finished": False,
    "lunch_packed": True,
    "practice_today": True
}

# Outer account check first; the schedule is only checked for a logged-in student.
if student["logged_in"]:
    print("Welcome back - here is today's plan")

    # Nested decision 1: homework depends on whether there is practice first
    if student["homework_finished"]:
        if student["practice_today"]:
            print("Homework done - go to practice, then dinner")
        else:
            print("Homework done - enjoy free time before dinner")
    else:
        if student["practice_today"]:
            print("Go to practice, then finish homework before dinner")
        else:
            print("Finish homework before any free time")

    # Nested decision 2: lunch plan depends on whether practice needs a snack
    if student["lunch_packed"]:
        if student["practice_today"]:
            print("Bring your packed lunch and an extra snack for practice")
        else:
            print("Bring your packed lunch")
    else:
        if student["practice_today"]:
            print("Buy lunch at school and pack a snack for practice")
        else:
            print("Buy lunch at school")
else:
    print("Log in first to see today's plan")
```

**Actual output:**

    Welcome back - here is today's plan
    Go to practice, then finish homework before dinner
    Bring your packed lunch and an extra snack for practice

It matches the prediction.


## Testing every meaningful outcome

Each test case below uses a different `student` dictionary. The dashboard code is the same as above, placed inside a procedure so it can run once per test case. These five cases reach all 9 recommendations.


```python
# CODE_RUNNER: Homework Hack - Test every dashboard outcome

def plan_day(student):
    if student["logged_in"]:
        print("Welcome back - here is today's plan")

        # Nested decision 1: homework depends on whether there is practice first
        if student["homework_finished"]:
            if student["practice_today"]:
                print("Homework done - go to practice, then dinner")
            else:
                print("Homework done - enjoy free time before dinner")
        else:
            if student["practice_today"]:
                print("Go to practice, then finish homework before dinner")
            else:
                print("Finish homework before any free time")

        # Nested decision 2: lunch plan depends on whether practice needs a snack
        if student["lunch_packed"]:
            if student["practice_today"]:
                print("Bring your packed lunch and an extra snack for practice")
            else:
                print("Bring your packed lunch")
        else:
            if student["practice_today"]:
                print("Buy lunch at school and pack a snack for practice")
            else:
                print("Buy lunch at school")
    else:
        print("Log in first to see today's plan")


test_cases = [
    ("Starter data", {"logged_in": True, "homework_finished": False, "lunch_packed": True, "practice_today": True}),
    ("Everything done, no practice, buying lunch", {"logged_in": True, "homework_finished": True, "lunch_packed": False, "practice_today": False}),
    ("Homework done, practice today, buying lunch", {"logged_in": True, "homework_finished": True, "lunch_packed": False, "practice_today": True}),
    ("Homework not done, no practice, lunch packed", {"logged_in": True, "homework_finished": False, "lunch_packed": True, "practice_today": False}),
    ("Logged out", {"logged_in": False, "homework_finished": True, "lunch_packed": True, "practice_today": True}),
]

for name, student in test_cases:
    print("Test:", name)
    plan_day(student)
    print()
```

**Actual output:**

| Test | logged_in | homework_finished | lunch_packed | practice_today | Homework message | Lunch message |
|---|---|---|---|---|---|---|
| Starter data | True | False | True | True | Go to practice, then finish homework before dinner | Bring your packed lunch and an extra snack for practice |
| Everything done, no practice, buying lunch | True | True | False | False | Homework done - enjoy free time before dinner | Buy lunch at school |
| Homework done, practice today, buying lunch | True | True | False | True | Homework done - go to practice, then dinner | Buy lunch at school and pack a snack for practice |
| Homework not done, no practice, lunch packed | True | False | True | False | Finish homework before any free time | Bring your packed lunch |
| Logged out | False | True | True | True | Log in first to see today's plan | (not checked) |

Coverage check:

- Homework messages: done + practice (test 3), done + no practice (test 2), not done + practice (test 1), not done + no practice (test 4). **All 4 reached.**
- Lunch messages: packed + practice (test 1), packed + no practice (test 4), buying + practice (test 3), buying + no practice (test 2). **All 4 reached.**
- Logged out (test 5) prints only `Log in first to see today's plan`, even though every other value is True. The inner checks never run.


## Why the inner checks depend on the outer decisions

The homework and lunch checks sit inside the `logged_in` branch because a dashboard should not show anyone's schedule until the account check passes; for a logged-out student, those questions are never even asked. Inside that branch, `practice_today` is nested under the homework and lunch answers because practice changes what each answer means: unfinished homework has to wait until after practice, and a packed lunch only needs an extra snack if there is practice. A compound condition like `homework_finished and practice_today` would treat every rule as one flat check, while nesting makes each later question depend on the path the earlier one chose.
