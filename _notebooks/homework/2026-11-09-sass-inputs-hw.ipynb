## SASS Inputs: OCS Input Grammar

Aaditya Prem (rex0dex). The idea of this lesson is the same as buttons and grids: don't style each field by hand. Every input gets the base class `ocs__input`, then one size modifier (`small`, `medium`, or `large`), plus `gradient` when the field should stand out. The theme handles the border, background, text color, and focus ring for me.

| Hack | What I did |
| --- | --- |
| Popcorn | Turned the dashed purple 400px search box into `ocs__input large gradient` and the made up `box` field into `ocs__input small` |
| Homework | Refactored the PVO signup form: Assistance Category is `large gradient`, Volunteer First Name is `small`, Volunteer Email is `medium`. No inline styles, no made up classes |

Each input also has a real `<label>` tied to it with `for` / `id`, so every field has a clear purpose even after the placeholder disappears. I used the stock `ocs__field` and `ocs__label` classes for that instead of writing my own spacing.

## Submission Safety Rules

What I followed on this page:

* No custom CSS, no `<style>` block, and no inline style attributes.
* No made up classes. The only classes are stock OCS ones: `ocs__input`, `small`, `medium`, `large`, `gradient`, plus `ocs__field`, `ocs__label`, and `ocs__card` for the labels and the form box.
* Every hack cell keeps `%%html` and its `UI_RUNNER` comment line.
* Each cell is only the final class based markup.

## Popcorn Hack: Refactor Two Inputs

**Starting code:** a search box with an inline `width: 400px` and `border: 3px dashed purple`, and a name field with the made up class `box`.

**My fix:**

* The search box is the main thing on the page, so it became `ocs__input large gradient`. `large` replaces the hardcoded width as the way to make it bigger, and `gradient` replaces the dashed purple border as the way to make it stand out.
* The first name field is short, so it became the compact plain `ocs__input small`.
* Each input sits in an `ocs__field` with an `ocs__label`, so the label and field line up with the theme spacing.

```python
%%html
<!-- UI_RUNNER: Inputs Popcorn -->
<div class="ocs__field">
  <label class="ocs__label" for="pvo-search">Search PVO resources</label>
  <input id="pvo-search" type="text" class="ocs__input large gradient" placeholder="Search PVO resources...">
</div>
<div class="ocs__field">
  <label class="ocs__label" for="popcorn-first-name">Volunteer first name</label>
  <input id="popcorn-first-name" type="text" class="ocs__input small" placeholder="Volunteer First Name">
</div>
```

## Homework Hack: PVO Volunteer Signup Form

**Starting code:** three inputs, each styled a different way:

* Assistance Category had `width: 500px` and a pink to purple `linear-gradient` background.
* Volunteer First Name used the made up class `tinybox`.
* Volunteer Email had `padding: 6px`.

**My fix:**

| Field | Before | After | Why |
| --- | --- | --- | --- |
| Assistance Category | inline width + gradient | `ocs__input large gradient` | It is the most important choice on the form, so it gets the biggest size and the theme gradient |
| Volunteer First Name | `tinybox` | `ocs__input small` | Short answer, so the compact size fits |
| Volunteer Email | inline padding | `ocs__input medium` | Standard size; `ocs__input` already gives every field the same padding |

A few other things I noticed:

* I kept `type="email"` on the email field, since the right input type is part of the lesson (the browser can check the format and phones show the email keyboard).
* The gradient is not pink to purple anymore. `gradient` uses the site's theme gradient, so it matches the rest of the page and still reads well in dark mode.
* I didn't replace `width: 500px` with anything. Inside `ocs__field` the input fills the width of the form, which is how the lesson says width should be handled (by the layout, not the input).
* The three fields are grouped in one `ocs__card` so they look like one signup form.

```python
%%html
<!-- UI_RUNNER: Inputs Homework -->
<div class="ocs__card">
  <div class="ocs__field">
    <label class="ocs__label" for="assistance-category">Assistance category</label>
    <input id="assistance-category" type="text" class="ocs__input large gradient" placeholder="Assistance Category">
  </div>
  <div class="ocs__field">
    <label class="ocs__label" for="volunteer-first-name">Volunteer first name</label>
    <input id="volunteer-first-name" type="text" class="ocs__input small" placeholder="Volunteer First Name">
  </div>
  <div class="ocs__field">
    <label class="ocs__label" for="volunteer-email">Volunteer email</label>
    <input id="volunteer-email" type="email" class="ocs__input medium" placeholder="Volunteer Email">
  </div>
</div>
```

## What I Learned

* HTML makes the input, SASS styles it, and JavaScript reads `input.value`. In AP CSP pseudocode the same thing is just `resource ← INPUT(...)` and `DISPLAY(...)`, so the form is the user interface part of the program (CRD-2.F).
* The size modifiers only change the font size. Width is a layout decision, so it belongs to the container (`ocs__field`), not the input.
* A placeholder is a hint, not a label. Once someone starts typing it disappears, so a real `<label>` is what keeps the field clear.
* Values from an input always come back as strings, so a number field would still need `Number(...)` before doing math with it.
