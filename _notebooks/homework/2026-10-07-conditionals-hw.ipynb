## 3.6 Conditionals

The scam-message trainer from our Friends of Poway Seniors capstone. A condition is a question that always comes back True or False, and every question needs all of its answers written down. Two failures never raise an error: a missing answer prints **nothing**, and an elif chain in the wrong order prints something **confidently wrong**.


## Popcorn Hack 1: the silent failure

**Step 1: what the original prints.** With `amount_requested = 0`, the condition `amount_requested > 100` is False and there is no `else`, so it prints **nothing**. No error, no warning, and the program finishes normally.

**Error or bug?** A **bug**. Python is happy: nothing crashed. The program just gave no answer for half its inputs. A scam trainer that says nothing looks exactly like one that says "this message is fine."

    amount_requested = 0
    if amount_requested > 100:
        print("This message is asking for a large amount of money.")

**Step 2: the fix.** One condition has 2 outcomes, so it needs 2 answers. Adding the `else` gives the missing outcome its own answer.


```python
# CODE_RUNNER: 3.6 Popcorn 1 - both outcomes print

amount_requested = 0

# 1 condition -> 2 outcomes -> 2 answers
if amount_requested > 100:
    print("This message is asking for a large amount of money.")
else:
    print("This message is not asking for a large amount of money.")
```

**Actual output:** `This message is not asking for a large amount of money.` With `amount_requested = 500` it prints the first answer instead, so both outcomes now print.


## Popcorn Hack 2: the chain in the wrong order

**Step 1: what the trainer says for a score of 9.** It prints **"Be a little careful, but this is probably fine."** That is the mildest warning, given to the worst possible message.

    if risk_score >= 3:
        print("Be a little careful, but this is probably fine.")
    elif risk_score >= 6:
        print("This looks risky. Check with someone you trust before replying.")
    elif risk_score >= 9:
        print("Do not reply. This is almost certainly a scam.")
    else:
        print("Nothing suspicious in this message.")

**Step 2: which branches can never run.** **Branch 2 (`elif risk_score >= 6`) and branch 3 (`elif risk_score >= 9`) are unreachable for every score.** Any score that is 6 or more is also 3 or more, so the first `if` catches it before Python ever asks the second question. An `elif` only runs when everything above it was False, and nothing at 6 or above can make `risk_score >= 3` False.

Counting does not catch this: there are 4 answers for 4 outcomes. The order makes two of them dead code.

**Step 3: the fix.** Order thresholds from highest to lowest, so each branch only sees the scores the branches above it turned down.


```python
# CODE_RUNNER: 3.6 Popcorn 2 - order fixed, highest threshold first

risk_score = 9

# 3 thresholds -> 4 outcomes -> 4 answers
if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
else:
    print("Nothing suspicious in this message.")
```

**Actual output:** `Do not reply. This is almost certainly a scam.` The worst score now gets the strongest warning.

**Checked every score from 0 to 10:**

| risk_score | Original (wrong order) | Fixed |
|---|---|---|
| 0, 1, 2 | Nothing suspicious | Nothing suspicious |
| 3, 4, 5 | Be a little careful | Be a little careful |
| 6, 7, 8 | Be a little careful | Check with someone you trust |
| 9, 10 | Be a little careful | Do not reply |

In the original, branches 2 and 3 never appear. In the fixed version, all four answers show up.


## Homework Hack: score a message, then advise on it

**Scoring.** One separate `if` per signal: +4 for gift cards, +3 for asking more than $100, +2 if the sender is not a known contact. They are separate `if`s, not an `elif` chain, because a message can trip **several** signals at once and every one of them should add to the score.

**Why the scoring `if`s need no `else`:** when a signal is not there, the right action is to add nothing, and leaving `risk_score` unchanged already does that. The "answer" to a False signal is simply that the score stays the same. The report chain afterwards is what makes sure something always prints.

**Reporting.** The highest possible score is 4 + 3 + 2 = 9, so the thresholds are 9, 6, and 3, ordered highest first so no branch is unreachable.


```python
# CODE_RUNNER: 3.6 Homework - score a message, then advise on it

message = "URGENT: send gift cards now to unlock your account"
amount_requested = 500
known_contact = False

risk_score = 0

# Scoring: one if per signal. No else needed: a missing signal adds nothing,
# and leaving risk_score unchanged is exactly the right answer for False.
if "gift card" in message.lower():
    risk_score = risk_score + 4
if amount_requested > 100:
    risk_score = risk_score + 3
if not known_contact:
    risk_score = risk_score + 2

print("Risk score:", risk_score)

# Reporting: highest threshold first so every branch is reachable.
# 3 thresholds -> 4 outcomes -> 4 answers
if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
else:
    print("Nothing suspicious in this message.")
```

**Actual output:** `Risk score: 9`, then `Do not reply. This is almost certainly a scam.` The starter message trips all three signals.


## Reachability proof

One set of inputs for each answer. Every branch has an input that reaches it, so none of them is dead code.

| Answer | message mentions gift cards? | amount_requested | known_contact | Score |
|---|---|---|---|---|
| Do not reply (score ≥ 9) | yes | 500 | False | 4 + 3 + 2 = **9** |
| Check with someone you trust (6–8) | yes | 500 | True | 4 + 3 = **7** |
| Be a little careful (3–5) | no | 500 | True | 3 = **3** |
| Nothing suspicious (0–2) | no | 0 | False | 2 = **2** |

All 8 combinations of the three signals give scores of 0, 2, 3, 4, 5, 6, 7, or 9, and every one of those lands in exactly one branch. The middle two rows run as cells below.


```python
# CODE_RUNNER: 3.6 Homework - reach the second answer (score 7)

message = "Please buy a gift card for the church raffle"
amount_requested = 150
known_contact = True

risk_score = 0

# Scoring: one if per signal. No else needed: a missing signal adds nothing,
# and leaving risk_score unchanged is exactly the right answer for False.
if "gift card" in message.lower():
    risk_score = risk_score + 4
if amount_requested > 100:
    risk_score = risk_score + 3
if not known_contact:
    risk_score = risk_score + 2

print("Risk score:", risk_score)

# Reporting: highest threshold first so every branch is reachable.
# 3 thresholds -> 4 outcomes -> 4 answers
if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
else:
    print("Nothing suspicious in this message.")
```

## Stretch: why `if` with no `else` is fine here

In Example A the missing `else` was the **only** answer the user would ever see, so a False condition meant the trainer said nothing. In the scoring step, a False signal correctly means "add nothing," and the reporting chain that follows always prints, so no outcome falls off the end.
