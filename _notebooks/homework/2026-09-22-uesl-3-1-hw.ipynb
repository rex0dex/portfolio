## Popcorn

**Prediction:** `stars` prints 12 and `checkpoint_stars` prints 8.

**Actual output:**

    12
    8

**Why they differ.** `checkpoint_stars = stars` copies the *value* 8 into a second name. It does not tie the two names together. So when `stars = stars + 4` runs, Python evaluates the right side first (8 + 4 = 12) and stores 12 in `stars` only. `checkpoint_stars` still holds the number it was handed at the moment of assignment, which is the whole point of a checkpoint — it remembers where you were, not where you are.


```python
# CODE_RUNNER: UESL 3.1 Popcorn - Keep the checkpoint

stars = 8
checkpoint_stars = stars   # copies the VALUE 8, not a link to the stars variable
stars = stars + 4          # only stars changes
print(stars)
print(checkpoint_stars)
```

## Homework

Game name and a reduced-motion setting, three level-star variables starting at 12, 18 and 10, then a bonus of 5 on level two.

**Prediction:** original total 40, stored total still 40 after the bonus, recalculated total 45.


```python
# CODE_RUNNER: UESL 3.1 Homework - Star counter with a bonus

game_title = "Star Trail: Poway Run"
reduced_motion = True

level_one_stars = 12
level_two_stars = 18
level_three_stars = 10

total_stars = level_one_stars + level_two_stars + level_three_stars

print(game_title)
print("Reduced motion:", reduced_motion)
print("Original total:", total_stars)

# Bonus of 5 on level two
bonus = 5
level_two_stars = level_two_stars + bonus

# The stored total does NOT change on its own
print("Stored total after bonus:", total_stars)

# It only changes when we recalculate it
total_stars = level_one_stars + level_two_stars + level_three_stars
print("Recalculated total:", total_stars)
```

**Actual output:**

    Star Trail: Poway Run
    Reduced motion: True
    Original total: 40
    Stored total after bonus: 40
    Recalculated total: 45

The middle line is the one worth staring at. Level two really did change from 18 to 23, but `total_stars` still printed 40, because it is a number that was calculated once and stored — not a formula that re-runs itself. Nothing updates until the last line recalculates it.

## Design Thinking

**Player need.** A player who gets motion sick, or who has trouble tracking fast movement, can't use a game that animates star counters flying across the screen. They don't need the game to be easier — they need it to hold still.

**My goal.** Make that a setting the player controls, stored in the game's state like any other value, instead of something hardcoded for everyone.

**A representation I considered.** I thought about storing motion as a speed number, like `preview_speed = 0.75`, and treating "reduced motion" as any value below 1. I chose a separate `reduced_motion` boolean instead, because the question the game actually asks is yes-or-no. Encoding that as a threshold on a float means every piece of code that checks it has to know the magic cutoff, and the day someone changes 0.75 to 0.8 the meaning silently shifts.

**What I prototyped.** The star counter above, with `reduced_motion` printed alongside the game name, so the setting is part of the state from the start rather than bolted on later.

**A test and what it changed.** Running the bonus twice — once with 5 and once with 0 — showed the stored total doesn't update itself. My first version printed the total only once, at the end, which hid the whole problem. I changed it to print the stored total *before* recalculating, because the gap between those two numbers is the thing being taught.

**Why reduced motion should be a player choice.** A game maker can't tell from the outside who needs it. Making it a stored setting means the game asks once and remembers, instead of each player having to work around a decision somebody else made for them. That's also exactly what a variable is for: a named value that can differ per player and change while the program runs.
