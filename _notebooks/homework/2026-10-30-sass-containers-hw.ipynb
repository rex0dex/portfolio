## SASS Containers HW

Theme: the Poway Scripps Rotary Parade. Both hacks take a broken starter page and rebuild it with only the OCS container grammar, so there is no CSS of my own anywhere in this notebook.

The nesting I followed every time:

- one `ocs__container` for the whole page (the outer boundary, max 900px and centered)
- one `ocs__card` inside it to group the content
- an `ocs__grid ocs__grid--standard cols-3` inside the card, made of `ocs__grid-cell` divs
- `ocs__table-wrap` + `ocs__table`, and an `ocs__callout`, also inside the card

No `style="..."`, no made-up classes like `page-wrap`, `box`, `wrap`, or `tile`.

## Popcorn Hack: Parade Invite Rescue

The starter had a `page-wrap` div with an inline `max-width`, a purple inline `h1`, and a hand-made `display: flex` row of three `box` divs. The flex row never wraps, so on a phone the three boxes get squished into one line.

What I changed:

- `page-wrap` became the one `ocs__container`
- everything went into one `ocs__card`, with the title as an `ocs__section-title` and a one-line `ocs__description`
- the flex row became `ocs__grid ocs__grid--standard cols-3` with three `ocs__grid-cell` divs, and the location cell gets `--accent` since that is the most important thing to know
- every inline `style` and the `box` class are gone

```python
%%html
<!-- UI_RUNNER: Containers Popcorn | show_code: true -->
<div class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">Poway Rotary Parade Invite</h2>
    <p class="ocs__description">You're invited to watch the Poway Scripps Rotary Parade roll down Poway Road this Saturday.</p>
    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell ocs__grid-cell--accent">Location: Poway Road</div>
      <div class="ocs__grid-cell">Start time: 9:00AM</div>
      <div class="ocs__grid-cell">What you get: an invaluable local and cultural experience!</div>
    </div>
  </div>
</div>
```

**Testing:** On a wide screen the three cells sit in one row of three. When I make the window narrower than 768px, the standard grid drops to 2 columns, so Location and Start time share the first row and "What you get" moves to the second row. Nothing gets squished or runs off the side, and the container keeps the whole invite centered.

## Homework Hack: Parade Hub Page

The starter was a `wrap` div with an inline `max-width`, a teal inline `h1`, a flex row of three `tile` divs, and a plain `border='1'` table with one row that just said "entry / open".

My rebuild, top to bottom, all inside one `ocs__container` and one `ocs__card`:

1. **Title + description:** `ocs__section-title` and a one-line `ocs__description`.
2. **Grid:** `ocs__grid ocs__grid--standard cols-3`. The first cell is `--header`, so it spans the full width as a label for the grid. Location is the `--accent` cell because it is the main thing people need. The rain plan is the `--muted` cell because it only matters if the weather turns bad. I used six regular cells under the header so the grid fills evenly at both sizes: 3 + 3 on a computer and 2 + 2 + 2 on a phone.
3. **Table:** the plain table became `ocs__table-wrap` around an `ocs__table`, with one header row (Event / Status) and two body rows. I turned the starter's "entry / open" row into a real "Float entries / Open" row.
4. **Callout:** one `ocs__callout` line at the end as the takeaway.

```python
%%html
<!-- UI_RUNNER: Containers Homework | show_code: true -->
<div class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">Poway Rotary Parade</h2>
    <p class="ocs__description">Everything you need for parade day on Poway Road, all in one place.</p>
    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell ocs__grid-cell--header">Parade Day at a Glance</div>
      <div class="ocs__grid-cell ocs__grid-cell--accent">Location: Poway Road</div>
      <div class="ocs__grid-cell">Start time: 9:00AM Saturday</div>
      <div class="ocs__grid-cell">Parade floats: local clubs and schools ride the route</div>
      <div class="ocs__grid-cell">Volunteers: greeters, setup crew, and cleanup team</div>
      <div class="ocs__grid-cell">Food drive: bring canned food to the Rotary booth</div>
      <div class="ocs__grid-cell ocs__grid-cell--muted">Rain plan: check this page the morning of the parade</div>
    </div>
    <div class="ocs__table-wrap">
      <table class="ocs__table">
        <thead>
          <tr><th>Event</th><th>Status</th></tr>
        </thead>
        <tbody>
          <tr><td>Float entries</td><td>Open</td></tr>
          <tr><td>Food drive</td><td>Collecting donations</td></tr>
        </tbody>
      </table>
    </div>
    <div class="ocs__callout">One container holds the whole Parade hub, and it looks the same on phones and computers with zero custom CSS.</div>
  </div>
</div>
```

**Testing narrow and wide:**

- **Wide (computer):** the container stops at 900px and stays centered. The header cell stretches across all three columns, then the six cells line up as two full rows of three. The table and callout stretch to the card's width under the grid.
- **Narrow (phone, under 768px):** the standard grid switches to 2 columns. The header still spans the full width, and the six cells become three even rows of two, so there is no lonely cell at the end. The `ocs__table-wrap` lets the table scroll sideways if it ever gets too wide instead of breaking the page.

**Cleanliness check:** one `ocs__container`, one `ocs__card`, no `style` attributes, no `<style>` block, and every class starts with `ocs__` (plus the `cols-3` modifier that the grid uses).

## What I Learned

The container is the outside layer and everything else goes inside it, kind of like nesting procedures in a program: the card groups things, the grid lays them out, and the cells hold the content. Because the styling all lives in the shared OCS SASS files, I didn't have to write any CSS. The responsive part (3 columns to 2) came for free just from using `ocs__grid--standard`, which is why the starter's hand-made flex row was worse even though it had more code.
