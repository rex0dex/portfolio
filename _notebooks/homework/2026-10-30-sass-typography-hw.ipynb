## Popcorn Hack

The bad code used a `<style>` block plus fake `div`/`span` titles. I swapped them for a real heading, a subheading, and a paragraph, and let the OCS classes handle the look instead of custom CSS.

```python
%%html
<!-- UI_RUNNER: Typography Popcorn Hack -->
<h2 class="ocs_section-title">Why Cats Purr</h2>
<h3>A Sign of Happiness</h3>
<p class="ocs_text">Cats usually purr when they feel safe, cozy, and ready for a nap.</p>
```

## Homework Hack

I deleted the whole `<style>` block and every custom class, then grouped everything in an `ocs_card` with an `h2` title and `h3` section headings. The fake numbered lines became one real `<ol>` with three `<li>` items, and the bold `span` became `<strong>` so the importance is in the HTML, not just the font.

```python
%%html
<!-- UI_RUNNER: Typography Homework -->
<div class="ocs_card">
  <h2 class="ocs_section-title">How to Pat a Cat</h2>
  <h3>Have you ever wanted to...</h3>
  <p class="ocs_text">pat a cat? Well good for you, we will give step by step instructions on how to pat a cat!</p>
  <p class="ocs_text"><strong>How to pat a cat:</strong></p>
  <ol>
    <li>Put out your hand</li>
    <li>Touch the cat</li>
    <li>Move your hand back and forth</li>
  </ol>
  <h3>Congrats! You pat a cat :D</h3>
</div>
```

## Takeaway

Semantic tags tell the browser and screen readers what each piece of text is, and the OCS classes give it the site's style, so I never have to hardcode font sizes.
