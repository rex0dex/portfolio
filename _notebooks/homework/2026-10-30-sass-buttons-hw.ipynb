## SASS Buttons Grammar HW

Aaditya Prem (rex0dex)

For this homework I built every button out of OCS button grammar only. There are no `<style>` blocks, no `style="..."` attributes, and no hex colors. Each action is a real `<button>` (does something on the page) or `<a>` (goes to another page), and the look comes from stacking classes from `_sass`: `ocs__btn` for the base, `pill` / `fill` / `outline` for shape and state, `accent` or `alert-green` / `alert-yellow` / `alert-red` for tone, and `ocs__links` to group them.

## Popcorn Hack #1: Add the Base Button Class

The starter link had an empty `class=""`, so it showed up as a plain text link. Adding `ocs__btn` gives it the standard OCS button look with zero CSS. I put it inside `ocs__links` so it gets the same spacing as the other groups.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 1 - Base Button Class | show_code: true -->
<div class="ocs__links">
  <a class="ocs__btn" href="https://www.mygoodbrain.org/resources">Mental Health Resource Directory</a>
</div>
```

## Popcorn Hack #2: Modifiers & Button Groups

1. Added `pill` and `fill` to the "Complete Daily Check-In" button, so it is now `ocs__btn pill accent fill` (rounded and filled with the accent color).
2. Changed the plain wrapper `<div>` into `<div class="ocs__links">` so both actions sit in one spaced row.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 2 - Modifiers and Button Groups | show_code: true -->
<div class="ocs__links">
  <button class="ocs__btn pill accent fill" type="button">Complete Daily Check-In</button>
  <a class="ocs__btn outline" href="https://www.mygoodbrain.org/resources">Save Draft Journal</a>
</div>
```

## Popcorn Hack #3: Status Tones & Wide Layout Bar

1. Switched the check-in tone from `alert-yellow` to `alert-green`, since finishing a check-in is a positive action.
2. Upgraded the container to the wide layout. `ocs__links--wide` is a modifier on top of `ocs__links` (it only widens the gap), so the container is `ocs__links ocs__links--wide`.
3. Added `fill` to Get Crisis Support so it is `ocs__btn alert-red fill` and has the strongest visual priority.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 3 - Status Tones and Wide Layout Bar | show_code: true -->
<div class="ocs__links ocs__links--wide">
  <button class="ocs__btn alert-green outline" type="button">Complete Check-In</button>
  <a class="ocs__btn alert-red fill" href="https://www.mygoodbrain.org/resources">Get Crisis Support</a>
</div>
```

## Homework Hack #1: Adaptive Resource Toolbar

The starter badges were three `<a>` tags with inline colors, padding, and border-radius. I removed every `style="..."` (and the flex wrapper styles), fixed the broken markdown-style `href` values, and mapped each category to its urgency:

| Category | Classes | Why |
| :--- | :--- | :--- |
| Urgent Support | `ocs__btn alert-red fill pill` | red + filled = most urgent |
| Self-Care | `ocs__btn alert-green fill pill` | green = positive, healthy habit |
| Community | `ocs__btn accent outline pill` | accent outline = calmer, optional |

All three sit inside one `ocs__links` container, and `pill` replaces the old `border-radius: 20px`.

```python
%%html
<!-- UI_RUNNER: Homework Hack 1 - Adaptive Resource Toolbar | show_code: true -->
<div class="ocs__links">
  <a class="ocs__btn alert-red fill pill" href="https://www.mygoodbrain.org/resources">Urgent Support</a>
  <a class="ocs__btn alert-green fill pill" href="https://www.mygoodbrain.org/resources">Self-Care</a>
  <a class="ocs__btn accent outline pill" href="https://www.mygoodbrain.org/resources">Community</a>
</div>
```

## Homework Hack #2: Primary vs. Secondary Action Priority

The starter panel used a hardcoded white card and teal/red hex colors. I took all of that out and let the classes set the hierarchy instead:

| Priority | Action | Classes |
| :--- | :--- | :--- |
| Primary (high) | Log Mood Entry | `ocs__btn accent fill` |
| Secondary (medium) | View Weekly Trends | `ocs__btn outline` |
| Alert | Crisis Hotline | `ocs__btn alert-red outline` |

Log Mood Entry stays a `<button>` because it is an action on this page, while the other two are `<a>` links to resource pages. The starter's third link said "Mental Health Blog", but the requirements call it "Crisis Hotline", so I renamed it to match. Only the primary action is filled, so your eye goes to it first; the other two are bordered, and the red tone still makes the hotline easy to find.

```python
%%html
<!-- UI_RUNNER: Homework Hack 2 - Primary vs Secondary Action Priority | show_code: true -->
<div class="ocs__links">
  <button class="ocs__btn accent fill" type="button">Log Mood Entry</button>
  <a class="ocs__btn outline" href="https://www.mygoodbrain.org/resources">View Weekly Trends</a>
  <a class="ocs__btn alert-red outline" href="https://www.mygoodbrain.org/resources">Crisis Hotline</a>
</div>
```

## What I Learned

- OCS classes work like procedures with parameters (AP CSP AAP-3.A): `ocs__btn` is the procedure, and `fill`, `accent`, or `alert-red` are the arguments that change the result without rewriting any CSS.
- Picking a tone is basically selection (AAP-2.C): urgent goes `alert-red fill`, positive goes `alert-green`, and everything else stays `accent` or plain.
- The base `ocs__btn` already draws a border, so `outline` mostly labels intent ("this is the secondary choice"), while `fill` is what actually makes a button stand out.
- `ocs__links` handles spacing and wrapping, so I never needed a custom flex container.
