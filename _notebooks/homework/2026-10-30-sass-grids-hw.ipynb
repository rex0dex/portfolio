## Popcorn Hack 1: Fix the Beaker Row

**Starting code:** an outer `<div style="display: flex;">` and three `<div class="box">` beakers. The inline style and the `box` class are both against the rules.

**My fix:**

* The outer div became `ocs__grid ocs__grid--standard`. That is the grid root plus one variant, and it gives 3 boxes per row (2 on phones).
* Each beaker became an `ocs__grid-cell`, so all three get the same panel, border, and hover from the theme.
* No `style` attribute and no `box` class are left.

````python
%%html
<!-- UI_RUNNER: Grids Popcorn 1 -->
<div class="ocs__grid ocs__grid--standard">
  <div class="ocs__grid-cell">Beaker A: Water</div>
  <div class="ocs__grid-cell">Beaker B: Salt Water</div>
  <div class="ocs__grid-cell">Beaker C: Vinegar</div>
</div>
````

## Popcorn Hack 2: Size the Beakers with `fr`

**Task:** all 4 beakers in one row, with the Control Beaker twice as wide as each of the others.

**How I picked the values:**

* 4 boxes in a row means 4 `fr` values, so the starter `1fr` (one box per row) had to become four values.
* The Control Beaker is last, so the last value is `2fr` and the rest are `1fr`.
* That is 5 shares total. Beakers 1 to 3 get 1/5 of the row each and the Control Beaker gets 2/5, which is exactly twice as wide.

**fr values used:** `1fr 1fr 1fr 2fr`

````python
%%html
<!-- UI_RUNNER: Grids Popcorn 2 -->
<div class="ocs__grid ocs__grid--standard" style="grid-template-columns: 1fr 1fr 1fr 2fr;">
  <div class="ocs__grid-cell">Beaker 1</div>
  <div class="ocs__grid-cell">Beaker 2</div>
  <div class="ocs__grid-cell">Beaker 3</div>
  <div class="ocs__grid-cell">Control Beaker</div>
</div>
````

| # | Question | My answer | Why |
| --- | --- | --- | --- |
| 1 | Make a standard grid show 4 boxes per row | **A.** `cols-4` | `cols-4` switches the standard grid to 4 columns, the same as `1fr 1fr 1fr 1fr` |
| 2 | What does `grid-template-columns: 2fr 1fr` make? | **B.** 2 boxes, the first twice as wide | Two values means two boxes, and 2 of the 3 shares go to the first one |
| 3 | Which cell type stretches across the whole row? | **C.** `ocs__grid-cell--header` | `--header` spans every column, while `--wide` only spans 2 |
| 4 | Which line breaks the rule? | **D.** `<div class="my-row" style="display:flex">` | `my-row` is a made up class and `display:flex` is an inline style |

## Homework Hack: Fix Both Grids

**Starting code:** a `my-row` div with `title-box`, `sample-card`, and `highlight` classes plus a `width: 100%` style, and a keypad built with `display: grid; grid-template-columns: repeat(4, 1fr)` and `key` / `orange` classes. All of that is hand made layout.

**Grid 1, the temperature trial:**

* `ocs__grid ocs__grid--standard`, because these are rows of readings.
* The title is `ocs__grid-cell--header`, so it stretches across the whole row on its own. That replaces `width: 100%`.
* Beaker 3 (31.6 °C) is way above the other two, so it is the outlier and gets `ocs__grid-cell--accent`. That replaces `highlight`.

**Grid 2, the keypad:**

* `ocs__grid ocs__grid--calculator`, which already gives 4 square keys per row. That replaces the `repeat(4, 1fr)` style.
* Every key is an `ocs__grid-cell`, and `Avg` gets `ocs__grid-cell--accent` so it stands out like the old orange key.

That is 8 boxes with `ocs__grid-cell`, one header, and two accent cells. The checklist below the grids runs the same checks as the lesson runner on my markup and marks each line PASS or FIX.

This is the same fix as I typed it into the lesson runner:

```js
let myHTML = [
  '<div class="ocs__grid ocs__grid--standard">',
  '  <div class="ocs__grid-cell ocs__grid-cell--header">Temperature Trial (°C)</div>',
  '  <div class="ocs__grid-cell">Beaker 1: 22.4</div>',
  '  <div class="ocs__grid-cell">Beaker 2: 22.9</div>',
  '  <div class="ocs__grid-cell ocs__grid-cell--accent">Beaker 3: 31.6</div>',
  '</div>',
  '<div class="ocs__grid ocs__grid--calculator">',
  '  <div class="ocs__grid-cell">7</div>',
  '  <div class="ocs__grid-cell">8</div>',
  '  <div class="ocs__grid-cell">9</div>',
  '  <div class="ocs__grid-cell ocs__grid-cell--accent">Avg</div>',
  '</div>'
].join('');
```

````python
%%html
<!-- UI_RUNNER: Grids Homework -->
<div id="grids-hw-lab">
  <div class="ocs__grid ocs__grid--standard">
    <div class="ocs__grid-cell ocs__grid-cell--header">Temperature Trial (°C)</div>
    <div class="ocs__grid-cell">Beaker 1: 22.4</div>
    <div class="ocs__grid-cell">Beaker 2: 22.9</div>
    <div class="ocs__grid-cell ocs__grid-cell--accent">Beaker 3: 31.6</div>
  </div>
  <div class="ocs__grid ocs__grid--calculator">
    <div class="ocs__grid-cell">7</div>
    <div class="ocs__grid-cell">8</div>
    <div class="ocs__grid-cell">9</div>
    <div class="ocs__grid-cell ocs__grid-cell--accent">Avg</div>
  </div>
</div>
<div id="grids-hw-checks"></div>
<script>
  // Same checks as the lesson's Homework runner, run on the grids above.
  var lab = document.getElementById("grids-hw-lab");
  var out = document.getElementById("grids-hw-checks");
  if (lab && out) {
    var myHTML = lab.innerHTML;
    var classList = [];
    (myHTML.match(/class="[^"]*"/g) || []).forEach(function (found) {
      classList = classList.concat(found.slice(7, -1).split(" ").filter(function (c) { return c !== ""; }));
    });
    var madeUp = classList.filter(function (c) { return c.indexOf("ocs__") !== 0 && c.indexOf("cols-") !== 0; });
    var count = function (name) { return classList.filter(function (c) { return c === name; }).length; };
    var line = function (label, ok) { return "<li>" + (ok ? "PASS  " : "FIX  ") + label + "</li>"; };
    out.innerHTML = "<ul>" +
      line("No inline style attributes", myHTML.indexOf("style=") === -1) +
      line("No made up classes" + (madeUp.length > 0 ? " (found: " + madeUp.join(", ") + ")" : ""), madeUp.length === 0) +
      line("Has an ocs__grid root", classList.indexOf("ocs__grid") !== -1) +
      line("Uses the standard variant for Grid 1", classList.indexOf("ocs__grid--standard") !== -1) +
      line("Uses the calculator variant for Grid 2", classList.indexOf("ocs__grid--calculator") !== -1) +
      line("Every box uses ocs__grid-cell", count("ocs__grid-cell") >= 8) +
      line("Title is a header cell", classList.indexOf("ocs__grid-cell--header") !== -1) +
      line("Has accent cells (outlier beaker and Avg key)", count("ocs__grid-cell--accent") >= 2) +
      "</ul>";
  }
</script>
````
