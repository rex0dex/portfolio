## 3.09 Developing Algorithms

Every algorithm is built from sequencing, selection and iteration. This lesson is about telling whether two algorithms really match. Algorithms that look different can be equivalent, and algorithms that look almost the same can give different results. You find out by testing below, at and above the limit, not by comparing how the code looks.


## Popcorn Hack 1 — Translate pseudocode to Python

**Prediction:** 3 + 5 + 2 + 4 = **14** hours, and 14 / 4 = **3.5** average. The average prints as `3.5` because `/` always gives a float.

Each pseudocode line maps across: `←` becomes `=`, `FOR EACH h IN hours` becomes `for h in hours:`, `LENGTH(hours)` becomes `len(hours)`, and `DISPLAY` becomes `print`. The total uses a `for` loop, not `sum()`.


```python
# CODE_RUNNER: 3.09 Popcorn 1 - Translate the pseudocode into Python

# Step 1: make the hours list
hours = [3, 5, 2, 4]

# Step 2: start total at 0
total = 0

# Step 3: FOR EACH -> for ... in ...: add each h to total
for h in hours:
    total = total + h

# Step 4: LENGTH(hours) -> len(hours): compute the average
average = total / len(hours)

# Step 5: print total, then average
print(total)
print(average)
```

**Actual output:** `14` then `3.5`. Matches the prediction.


## Popcorn Hack 2 — Predict, then fix

**Prediction before running the original:** `Full workshops: 1`. With `count > capacity`, Self-Defense at exactly 12 is not greater than 12, so only Fitness (15) counts.

**Actual output of the original:** `Full workshops: 1`. It was wrong, with no error.

**The fix:** `>` becomes `>=`. "Full" means 12 **or more**, and the only input where `>` and `>=` disagree is exactly 12. That's the boundary this bug hides in.


```python
# CODE_RUNNER: 3.09 Popcorn 2 - Predict, then fix the full-workshop counter

# The data: Self-Defense, Parenting Skills, Fitness, Self-Sufficiency
signed_up = [12, 7, 15, 9]
capacity = 12

# Count workshops that are full: 12 OR MORE, so >= (was > before the fix)
full = 0
for count in signed_up:
    if count >= capacity:
        full = full + 1

# Report
print("Full workshops:", full)

# My prediction before running the original (with >): 1
# After fixing > to >=: 2
```

**Actual output after the fix:** `Full workshops: 2`, which is Self-Defense and Fitness.


## Popcorn Hack 3 — Modify the counter into a collector

**Prediction:** Parenting Skills (7) and Self-Sufficiency (9) are under 12. Self-Defense is at exactly 12, so it is **not** open. That gives `2` and `['Parenting Skills', 'Self-Sufficiency']`.

This is **modifying** an algorithm. The loop and the `if` are the same as the counter. Only what gets tracked changes: `count = count + 1` becomes `open_workshops.append(...)`.


```python
# CODE_RUNNER: 3.09 Popcorn 3 - Modify the counter into a collector

# Parallel lists: signed_up[0] goes with workshops[0], and so on
workshops = ["Self-Defense", "Parenting Skills", "Fitness", "Self-Sufficiency"]
signed_up = [12, 7, 15, 9]
capacity = 12

# COLLECT names instead of counting
open_workshops = []

# Keep a workshop only if it has fewer than 12 signed up
for i in range(len(workshops)):
    if signed_up[i] < capacity:
        open_workshops.append(workshops[i])

# Print how many there are, then the list
print("Workshops with open seats:", len(open_workshops))
print(open_workshops)
```

**Actual output:** `Workshops with open seats: 2` then `['Parenting Skills', 'Self-Sufficiency']`. Self-Defense is correctly left out.

## Homework Hack — Supply closet restock report

This homework **combines** three patterns I already trust:

1. **Collect:** keep each item where `in_stock < target`, along with how many it needs.
2. **Sum:** total the shortages with a loop, not `sum()`.
3. **Find the max:** track the *position* of the biggest shortage, so the item's name comes along with it.

**Prediction.** Toothbrushes 40/30 and Notebooks 30/30 are fine. Blankets 8/20 needs **12**. Shampoo 5/15 needs **10**. Socks 25/25 is exactly on target, and `<` is strict, so Socks is **not** short. That's the boundary again. Total 12 + 10 = **22**. Most urgent: **Blankets (12)**.


```python
# CODE_RUNNER: 3.09 Homework - Safe Passage Heals supply restock report

# The closet: parallel lists, so index i describes the same item in all three
items    = ["Toothbrushes", "Blankets", "Socks", "Shampoo", "Notebooks"]
in_stock = [40, 8, 25, 5, 30]      # in the closet now
target   = [30, 20, 25, 15, 30]    # what staff want on hand

# Step 1: COLLECT items below target, and how many of each are needed
# (< not <=, so an item sitting exactly at target like Socks is NOT short)
short_items = []
short_amounts = []
for i in range(len(items)):
    if in_stock[i] < target[i]:
        short_items.append(items[i])
        short_amounts.append(target[i] - in_stock[i])

# Step 2: SUM the shortages with a loop (no sum())
total_needed = 0
for amount in short_amounts:
    total_needed = total_needed + amount

# Step 3: FIND THE MAX shortage, tracking its position so the name comes with it
# (only when something is short; short_amounts[0] would crash on an empty list)
if len(short_amounts) > 0:
    max_index = 0
    for i in range(len(short_amounts)):
        if short_amounts[i] > short_amounts[max_index]:
            max_index = i

# Step 4: COMBINE: "Fully stocked!" if nothing is short, otherwise the report
if len(short_items) == 0:
    print("Fully stocked!")
else:
    print("Restock needed:")
    for i in range(len(short_items)):
        print(short_items[i] + ": need " + str(short_amounts[i]))
    print("Total items needed: " + str(total_needed))
    print("Most urgent: " + short_items[max_index] + " (short by " + str(short_amounts[max_index]) + ")")
```

**Actual output:**

    Restock needed:
    Blankets: need 12
    Shampoo: need 10
    Total items needed: 22
    Most urgent: Blankets (short by 12)

This matches the expected output exactly. Socks is correctly left out.

**Testing the other branch.** The rubric asks for the "Fully stocked!" case too, so the next cell starts fresh with a closet where every item is at or above target.


```python
# CODE_RUNNER: 3.09 Homework - Test: a fully stocked closet

# Every item at or above target (Socks and Notebooks sit exactly AT target)
items    = ["Toothbrushes", "Blankets", "Socks", "Shampoo", "Notebooks"]
in_stock = [40, 20, 25, 18, 30]
target   = [30, 20, 25, 15, 30]

# Step 1: COLLECT items below target
short_items = []
short_amounts = []
for i in range(len(items)):
    if in_stock[i] < target[i]:
        short_items.append(items[i])
        short_amounts.append(target[i] - in_stock[i])

# Step 2: SUM the shortages with a loop
total_needed = 0
for amount in short_amounts:
    total_needed = total_needed + amount

# Step 3: FIND THE MAX, only when something is short
if len(short_amounts) > 0:
    max_index = 0
    for i in range(len(short_amounts)):
        if short_amounts[i] > short_amounts[max_index]:
            max_index = i

# Step 4: COMBINE
if len(short_items) == 0:
    print("Fully stocked!")
else:
    print("Restock needed:")
    for i in range(len(short_items)):
        print(short_items[i] + ": need " + str(short_amounts[i]))
    print("Total items needed: " + str(total_needed))
    print("Most urgent: " + short_items[max_index] + " (short by " + str(short_amounts[max_index]) + ")")
```

**Actual output:** `Fully stocked!` No crash.

**Why Step 3 has a guard.** Find-the-max starts from `short_amounts[0]`. On a fully stocked closet that list is empty, so without the `if len(short_amounts) > 0` check, this test would crash with `IndexError` instead of printing the message. The main test never reaches that path. Only this test does.
