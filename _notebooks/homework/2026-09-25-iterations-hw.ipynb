## Popcorn Hack — wildfire dispatch scanner

The blanks were the two cutoffs, the loop variable, and the two comparisons. What I picked and why:

- **`safe_cutoff = 3`.** Alerts of 1 and 2 are noise. Anything at 3 or above gets a crew.
- **`critical_cutoff = 9`.** High enough that the scan does not stop on a merely serious alert, low enough that a 9 halts it.

**`continue` versus `break` is the whole hack.** `continue` abandons *this* alert and moves to the next one — the scan keeps going. `break` abandons *the loop* — nothing after it is ever looked at. Skipping one value and stopping the scan are completely different, and swapping them is the mistake the hack is built to catch.

The counters start at 0 because nothing has been processed yet. If `total_action_score` started at anything else, every total after it would be wrong by that amount and nothing would warn you.

Expected: 2 safe zones skipped, 4 action alerts, total score 22, critical found.


```python
# CODE_RUNNER: Popcorn Hack - wildfire dispatch scanner

# Theme: The SFSRC team scans wildfire alert levels from different zones.
# Goal: skip safe zones, count response zones, total their alert scores, and stop at the first critical zone.

alerts = [1, 4, 2, 6, 3, 9, 5]

# Alerts below this number are safe and get skipped.
safe_cutoff = 3

# The loop stops at this number or higher.
critical_cutoff = 9

# Counters and totals start at 0 because no alerts have been processed yet.
action_count = 0
safe_count = 0
total_action_score = 0
critical_found = False

# Loop through the alert list.
for alert in alerts:
    # Safe zone: count it, say so, and skip the rest of this pass.
    if alert < safe_cutoff:
        safe_count += 1
        print("Zone alert " + str(alert) + ": safe, skipping")
        continue

    # Anything that gets here needs a crew.
    action_count += 1
    total_action_score += alert
    print("Dispatch responding to alert " + str(alert))

    # Critical zone: stop the whole scan so the team can focus.
    if alert >= critical_cutoff:
        critical_found = True
        print("Critical SFSRC alert found at " + str(alert))
        break

print("--- SFSRC scan summary ---")
print("Safe zones skipped: " + str(safe_count))
print("Action alerts: " + str(action_count))
print("Total action score: " + str(total_action_score))
print("Critical found: " + str(critical_found))
```

**The last alert never gets looked at.** `alerts` has 7 values but the scan only reaches 6 of them — the `5` after the `9` is never inspected, because `break` ended the loop. That is the intended behaviour here, but it is worth saying out loud: after a `break`, the counts describe *what was scanned*, not *what exists*. A summary that claims otherwise would be lying.

Question 1 of the MCQ was *"Which loop is best when you already have a list of values to inspect?"* — **A, the `for` loop**. A `for` loop traverses a known sequence and handles the stopping for you. A `while` loop would also work, but you would have to create and update the index by hand, which is exactly where off-by-one and infinite-loop bugs come from. `break` and `continue` are not loops at all — they are controls used *inside* one.

## Homework Hack — scam-message triage for Friends of Poway Seniors

**My monitoring system.** Our capstone is a scam-defense trainer for Friends of Poway Seniors, so my checker scans an inbox of messages that have each been scored 1–10 for how scam-like they look, and decides which ones a volunteer needs to read.

Meeting the requirements:

| Requirement | What I used |
|---|---|
| At least 8 values | 8 risk scores |
| 2 thresholds | `review_cutoff = 4`, `critical_cutoff = 9` |
| `for` loop | traverses the score list |
| `continue` | skips messages below the review cutoff |
| `break` | stops at the first critical message |
| 3+ tracked summary values | checked, skipped, flagged, total score, critical found — 5 |
| Theme in the names | `message_risk_scores`, `review_cutoff`, `flagged_count` |

**Why stopping early makes sense here, and not everywhere.** A critical message means a volunteer needs to act now, so the scan halts and hands off rather than finishing the queue. That is a real design choice, not just a way to use `break` — and its cost is that the rest of the inbox goes unread until someone restarts the scan.

Expected: 7 checked, 3 skipped, 4 flagged, total 27, critical found.


```python
# CODE_RUNNER: Homework Hack - scam-message triage checker

# Theme: Friends of Poway Seniors scam-defense trainer.
# Each incoming message has been scored 1-10 for how scam-like it looks.
# Goal: skip low-risk messages, flag the rest for a volunteer, and stop at the first critical one.

message_risk_scores = [1, 5, 2, 7, 3, 6, 9, 4]

# Below this, the message is not worth a volunteer's time.
review_cutoff = 4

# At or above this, a volunteer needs to act immediately.
critical_cutoff = 9

# Summary values, all starting from nothing processed.
checked_count = 0
flagged_count = 0
skipped_count = 0
total_risk_score = 0
critical_found = False

for score in message_risk_scores:
    checked_count += 1

    # Low risk: note it and move to the next message.
    if score < review_cutoff:
        skipped_count += 1
        print("Risk " + str(score) + ": below review threshold, skipping")
        continue

    # Everything that gets here goes to a volunteer.
    flagged_count += 1
    total_risk_score += score
    print("Risk " + str(score) + ": flagged for volunteer review")

    # Critical: stop the scan and hand off now.
    if score >= critical_cutoff:
        critical_found = True
        print("Risk " + str(score) + ": critical, stopping the scan")
        break

print("--- Friends of Poway Seniors inbox scan ---")
print("Messages in the queue: " + str(len(message_risk_scores)))
print("Messages checked: " + str(checked_count))
print("Skipped as low risk: " + str(skipped_count))
print("Flagged for volunteer review: " + str(flagged_count))
print("Total risk score of flagged messages: " + str(total_risk_score))
print("Critical message found: " + str(critical_found))
```

## What I actually took away
