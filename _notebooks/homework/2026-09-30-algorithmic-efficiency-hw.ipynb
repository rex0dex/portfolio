## Library 3.17 Algorithmic Efficiency

A librarian wants to find books quickly as the catalog grows. Two searches can both be correct and still do very different amounts of work, so this homework compares them by counting checks instead of counting lines of code.


## Popcorn

**Warm-up first — find book 15 in `[3, 6, 9, 12, 15, 18, 21]`.** Linear search checks 3, 6, 9, 12, 15: **5 checks**. Binary search checks 12, then 18, then 15: **3 checks**. Binary saves 2. For book 3, the result flips: linear finds it on the **1st** check, but binary still takes **3** (12, then 6, then 3). So binary search does not win on every single target. It wins on how the worst case grows.

**Checkpoint B — expand the catalog.**

**Prediction before running.** For 8 books, linear search's worst case checks every book, so **8**. Binary search halves 8 → 4 → 2 → 1, which is 3 halvings plus the final check, so **4**. If n doubles to 16, I predicted that linear doubles to **16** and binary only goes up by one, to **5**.

**Edited calculator:** `n` changed from `8` to `16`. Both versions are below, each starting fresh.


```python
# CODE_RUNNER: 3.17 Popcorn - Expand the catalog (n = 8)
import math

n = 8
linear_checks = n
binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1
print(n)
print(linear_checks)
print(binary_checks)
```

```python
# CODE_RUNNER: 3.17 Popcorn - Expand the catalog (n changed to 16)
import math

n = 16
linear_checks = n
binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1
print(n)
print(linear_checks)
print(binary_checks)
```

**Actual output:** `8, 8, 4`, then `16, 16, 5`. Matches the prediction.

**Which count doubles, which goes up by one.** Linear checks doubled from 8 to 16, because every new book is one more book that might have to be inspected. Binary checks only went from 4 to 5, because doubling the catalog adds just one more halving.

**Recommendation.** For a sorted catalog, use binary search. The gap only grows with the catalog: at 16 books it saves 11 checks, and at 1,000 books it saves 990.

## Homework

**Adapting Example C for 12 books.** `visits` is scanning each book once, so it equals n. `pairs` is comparing every book with every book, including itself and in both orders, so it equals n × n.

**Prediction:** 12 books, 12 visits, 144 pairs. Doubled to 24: 24 books, 24 visits, 576 pairs.


```python
# CODE_RUNNER: 3.17 Homework - Advise the librarian (12 books)

n = 12
visits = n
pairs = n * n
print("Books in catalog: " + str(n))
print("Visits to scan every book: " + str(visits))
print("Ordered pairs to compare every book: " + str(pairs))
```

```python
# CODE_RUNNER: 3.17 Homework - Double the catalog (24 books)

n = 24
visits = n
pairs = n * n
print("Books in catalog: " + str(n))
print("Visits to scan every book: " + str(visits))
print("Ordered pairs to compare every book: " + str(pairs))
```

**Actual output:** `12, 12, 144`, then `24, 24, 576`. Matches the prediction.

**Why the growth is different.** Visits are **linear**: doubling the books from 12 to 24 doubles the visits from 12 to 24. Pairs are **quadratic**: doubling the books quadruples the pairs, 144 × 4 = 576, because each of the 24 books is now compared against 24 books instead of 12. That quadratic task gets expensive fast, but it can't be swapped for a scan if pairs are what the librarian actually needs.

**Recommendation for 1,000 books** (checkpoint B's calculator, below).


```python
# CODE_RUNNER: 3.17 Homework - Recommend a search for 1,000 books
import math

n = 1000
linear_checks = n
binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1
print("Books in catalog: " + str(n))
print("Linear search, worst case: " + str(linear_checks) + " checks")
print("Binary search, worst case: " + str(binary_checks) + " checks")
```

**Actual output:** 1,000 books, 1,000 linear checks, 10 binary checks.

**My recommendation.** For **repeated searches of a catalog that is already sorted**, use binary search. In the worst case it needs 10 checks instead of 1,000, and that saving happens again on every search.

**If the catalog is unsorted,** binary search gives wrong answers until the catalog is sorted, and sorting costs work of its own. For a single search, just doing a linear search can be cheaper. The more times the librarian searches, the more worthwhile it is to sort once.

**If I have to store an extra sorted copy,** the searches get faster but memory use goes up, because 1,000 entries are stored twice. That is a time vs. space tradeoff: fewer checks per search in exchange for more storage.

**Connection to the warm-up.** Better worst-case growth does not mean fewer checks on every target. Book 3 took 1 linear check and 3 binary checks. Correctness, one particular case, and growth are three separate questions.

```python
# CODE_RUNNER: 3.17 Tests - Empty catalog (0 books)
import math

n = 0
visits = n
pairs = n * n
binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1
print("Books in catalog: " + str(n))
print("Visits to scan every book: " + str(visits))
print("Ordered pairs to compare every book: " + str(pairs))
print("Binary search, worst case: " + str(binary_checks) + " checks")
```

**What the empty test proves.** Visits and pairs handle 0 without any special case, since 0 and 0 × 0 are both 0. Binary search is different: `math.log2(0)` is undefined and crashes with `ValueError: math domain error`. The `0 if n == 0 else ...` guard is the only reason this test passes. The main path never touches that guard, so only the empty catalog shows whether it's there.


## Design Thinking

**Librarian need.** Find a book quickly even when the catalog has thousands of entries, and get advice that doesn't require understanding the formulas.

**Goal.** Compare two correct searches by the checks they do and how that count grows, then recommend one.

**Methods considered.** Linear search works on any catalog and needs no preparation, but its worst case grows 1 check per book. Binary search grows by only 1 check each time the catalog doubles, but it needs the catalog sorted and needs direct access to the middle entry.

**Prototype.** Checkpoint B's calculator, reused for 1,000 books, plus Example C adapted with labeled output. Every line says what it is counting ("Visits to scan every book: 12") instead of printing bare numbers, so the librarian can read the result without knowing that 144 means 12 × 12.

**Actual test and revision.** The lesson's calculator prints bare numbers, and for 1,000 books that would be `1000`, `1000`, `10`, with two identical 1,000s and no way to tell the book count from the check count. My revision was to label every output line. The empty-catalog test then checked the one edge case the main path never reaches: without the `n == 0` guard, `math.log2(0)` raises `ValueError: math domain error`, so the guard stays.
