## Popcorn Hack: Fill in the Blanks

1. The **checkbox** on a toggle keeps track of whether it is on or off.
2. The **slider** is the visible part of the toggle that changes color when it is turned on.
3. The `.switch` class controls the toggle's position, size, padding, and **alignment**.
4. The **border-radius** property makes the slider have rounded edges.
5. The `&:before` class dictates the **white circle** of the toggle.

In the new grammar the slider is `ocs__toggle-track` and the circle is its `::before`, which slides over when the input is `:checked`.

## Find the Bug

The starter toggle had `<span class="">`, so there was nothing to draw the switch. Adding `ocs__toggle-track` brings the track back, and the runner printed **Toggle works!** I also added `aria-hidden="true"` because the track is only decoration, and the real state is the checkbox.

````python
%%html
<!-- UI_RUNNER: Find the Bug - Missing Track Class | show_code: true -->
<label class="ocs__toggle ocs__toggle--green">
  <input class="ocs__toggle-input" type="checkbox">
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Enable Notifications</span>
</label>
````

## Homework: Study Session Panel

I made a panel I would actually use before studying. Each toggle has its own label and its own job:

| Toggle | Classes | What it does when ON |
| :--- | :--- | :--- |
| Show Study Tips | `ocs__toggle ocs__toggle--green` | Shows a list of study tips |
| Show Today's Agenda | `ocs__toggle ocs__toggle--green` | Shows my plan for today |
| Focus Mode | `ocs__toggle` | Hides the distractions list |
| Break Reminder | `ocs__toggle ocs__toggle--green` | Changes the reminder text to a 25 minute timer plan |

How it covers each requirement:

- **At least 3 toggles:** there are 4, all using the `ocs__toggle` classes, each with its own label and purpose.
- **Check on/off with JavaScript:** `updatePage()` reads `.checked` on every toggle. The Toggle Status list says ON or OFF for each one, and each toggle shows or hides (or rewrites) part of the page.
- **Count of toggles on:** the `<output aria-live="polite">` line counts how many are on, and it updates the moment you flip one because it runs on the `change` event.
- **Comments:** the HTML and JavaScript have comments about what I learned and what confused me at first.
- **Above and beyond:** every toggle is connected to a real feature, and when all 4 are on the panel says I am ready to study.

Focus Mode is the one I liked most, because it is the opposite of the others: ON means *hide* something, so its line in the code does not use `!`.

````python
%%html
<!-- UI_RUNNER: Toggles Homework - Study Session Panel | show_code: true -->
<div id="study-panel">
  <p><strong>Study Session Panel</strong></p>

  <!-- Every toggle is the same 4 parts: the label wraps everything, so clicking the words also flips the switch. -->
  <!-- The checkbox is invisible, but it is the part that actually stores ON or OFF. The track is just what I see. -->
  <div>
    <label class="ocs__toggle ocs__toggle--green">
      <input id="tips-toggle" class="ocs__toggle-input" type="checkbox">
      <span class="ocs__toggle-track" aria-hidden="true"></span>
      <span class="ocs__toggle-label">Show Study Tips</span>
    </label>
  </div>
  <div>
    <label class="ocs__toggle ocs__toggle--green">
      <input id="agenda-toggle" class="ocs__toggle-input" type="checkbox">
      <span class="ocs__toggle-track" aria-hidden="true"></span>
      <span class="ocs__toggle-label">Show Today's Agenda</span>
    </label>
  </div>
  <div>
    <!-- No --green here on purpose, so Focus Mode turns on in the theme accent color and looks different from the rest. -->
    <label class="ocs__toggle">
      <input id="focus-toggle" class="ocs__toggle-input" type="checkbox">
      <span class="ocs__toggle-track" aria-hidden="true"></span>
      <span class="ocs__toggle-label">Focus Mode</span>
    </label>
  </div>
  <div>
    <label class="ocs__toggle ocs__toggle--green">
      <input id="break-toggle" class="ocs__toggle-input" type="checkbox">
      <span class="ocs__toggle-track" aria-hidden="true"></span>
      <span class="ocs__toggle-label">Break Reminder</span>
    </label>
  </div>

  <!-- aria-live="polite" makes screen readers read the new number after it changes, without interrupting. -->
  <p>Toggles on: <output id="on-count" aria-live="polite">0</output> of 4</p>
  <p id="ready-text"></p>

  <p><strong>Toggle Status</strong></p>
  <ul id="toggle-status"></ul>

  <!-- The features below start hidden with the hidden attribute, so no CSS is needed to show or hide them. -->
  <ul id="tips-box" hidden>
    <li>Do practice problems before rereading notes.</li>
    <li>Explain the topic out loud like you are teaching it.</li>
    <li>Put your phone in another room.</li>
  </ul>
  <p id="agenda-box" hidden>Today: finish the SASS Toggles HW, review conditionals, and do 10 practice MCQs.</p>
  <p id="distraction-box">Distractions: phone, YouTube tab, group chat.</p>
  <p id="break-text">Break reminder is off.</p>
</div>

<script>
  // I wrapped the code in a function so my variable names do not clash with other scripts on the page.
  (function () {
    const panel = document.getElementById("study-panel");
    if (!panel) return;

    // I only grab toggles inside my panel. The lesson's risk checker used document.querySelectorAll,
    // which would also count the other toggles higher up on this page.
    const toggles = panel.querySelectorAll(".ocs__toggle-input");

    const tipsToggle = document.getElementById("tips-toggle");
    const agendaToggle = document.getElementById("agenda-toggle");
    const focusToggle = document.getElementById("focus-toggle");
    const breakToggle = document.getElementById("break-toggle");

    const countOut = document.getElementById("on-count");
    const readyText = document.getElementById("ready-text");
    const statusList = document.getElementById("toggle-status");

    function updatePage() {
      // .checked is a boolean (true or false), so I can use it straight as the condition.
      // hidden = !checked means: toggle ON -> not hidden, toggle OFF -> hidden.
      document.getElementById("tips-box").hidden = !tipsToggle.checked;
      document.getElementById("agenda-box").hidden = !agendaToggle.checked;

      // Focus Mode is backwards from the others: ON should HIDE the distractions, so there is no ! here.
      // This took me a second to figure out because I copied the line above at first and it did the opposite.
      document.getElementById("distraction-box").hidden = focusToggle.checked;

      // This one changes text instead of hiding something. It is the IF / ELSE from the lesson's College Board section.
      if (breakToggle.checked) {
        document.getElementById("break-text").textContent = "Break reminder: study 25 minutes, then take a 5 minute break.";
      } else {
        document.getElementById("break-text").textContent = "Break reminder is off.";
      }

      // Count how many toggles are on, and rebuild the status list at the same time.
      let onCount = 0;
      statusList.textContent = "";
      toggles.forEach(function (toggle) {
        if (toggle.checked) {
          onCount += 1;
        }
        // I read the words from the toggle's own ocs__toggle-label, so if I rename a toggle the list updates too.
        const name = toggle.parentElement.querySelector(".ocs__toggle-label").textContent;
        const item = document.createElement("li");
        item.textContent = name + ": " + (toggle.checked ? "ON" : "OFF");
        statusList.appendChild(item);
      });
      countOut.textContent = onCount;

      // Selection on the count: only when every toggle is on do I get the ready message.
      if (onCount === toggles.length) {
        readyText.textContent = "Everything is on. Ready to study!";
      } else {
        readyText.textContent = "Turn on all " + toggles.length + " toggles to fully set up your session.";
      }
    }

    // The "change" event fires every time a toggle flips, so the count updates right away with no button to click.
    toggles.forEach(function (toggle) {
      toggle.addEventListener("change", updatePage);
    });

    // Run once at the start so the status list and count match the page before anything is clicked.
    updatePage();
  })();
</script>
````

## What I Learned

- A toggle is really just a checkbox in disguise. The OCS classes hide the checkbox and draw the track, so I get a switch without writing any CSS.
- `.checked` is a boolean, which makes a toggle a direct example of selection (AP CSP AAP-2.H): IF it is on, do one thing, ELSE do the other.
- Counting the toggles that are on is iteration (AAP-2.K): loop through the list and add 1 each time the condition is true, the same pattern as the lesson's risk checker.
- The `change` event is what makes the page interactive. Without it, the count would only update when something else told it to.
